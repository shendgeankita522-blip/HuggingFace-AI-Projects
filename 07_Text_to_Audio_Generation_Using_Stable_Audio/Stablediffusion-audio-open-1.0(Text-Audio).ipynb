```python
import torch
import soundfile as sf
from diffusers import StableAudioPipeline


model_id = "stabilityai/stable-audio-open-1.0"


# Load model
pipe = StableAudioPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")


# Bird sound prompt
prompt = (
    "A realistic field recording of birds chirping and singing "
    "in a peaceful forest during early morning with natural "
    "outdoor ambience."
)

negative_prompt = "Low quality, distorted audio, noise"


generator = torch.Generator("cuda").manual_seed(0)

# Generate audio
audio = pipe(
    prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=100,
    audio_start_in_s=0.0,
    audio_end_in_s=10.0,
    generator=generator
).audios[0]


# Convert CUDA float16 tensor → CPU float32 numpy
audio = audio.cpu().numpy().astype("float32")


# Save audio
sf.write(
    "birds_chirping.wav",
    audio.T,
    44100
)


print("Bird audio generated successfully!")
```
